# **Maximizar el valor del proyecto**

**La pregunta clave es:**

¿Cómo logramos que el proyecto genere el mayor valor posible al final del día?

**Para responderla, debemos analizar:**

CAPEX: gastos de capital, es decir, inversiones que se realizan para adquirir o desarrollar activos que tendrán un beneficio a largo plazo.


OPEX: gastos operativos, es decir, los costos necesarios para mantener la operación, como suscripciones, servicios o licencias.




# **Métricas de viabilidad financiera**

Para saber si un proyecto es financieramente viable, podemos utilizar diferentes indicadores:

* **VAN (Valor Actual Neto):** permite calcular cuánto valor genera realmente el proyecto en el presente, considerando los ingresos y costos futuros y aplicando una tasa de descuento.


* **TIR (Tasa Interna de Retorno):** muestra la tasa de rentabilidad que puede generar el proyecto.


* **ROI (Retorno de la Inversión):** permite comparar cuánto se obtiene frente a lo que se invirtió.


* **Periodo de recuperación:** indica cuánto tiempo se necesita para recuperar la inversión inicial.

Estimación de costos y línea base presupuestal
1. Estimación de costos

La estimación de costos consiste en calcular cuánto dinero se necesita para realizar cada actividad del proyecto, teniendo en cuenta los recursos necesarios.

Se busca obtener una estimación realista que permita saber:

Qué actividades se van a realizar.

Qué recursos se necesitan.

Cuánto cuesta cada recurso.

Cuánto cuesta cada actividad.

Cuánto costará el proyecto en total.

Cómo utilizar los recursos para obtener el máximo valor posible.

En otras palabras:

Estimar es calcular cuánto necesitamos gastar para poder cumplir lo que prometimos entregar.

2. Costo de la calidad

La calidad también tiene un costo. Por eso, el proyecto debe encontrar un equilibrio entre invertir para prevenir problemas y asumir el costo de corregirlos.

Existen cuatro conceptos importantes:

Prevención: invertir para evitar que ocurran errores.

Evaluación: revisar y comprobar que el trabajo cumple con los requisitos.

Fallos internos: errores encontrados antes de entregar el producto al cliente.

Fallos externos: errores encontrados después de entregar el producto al cliente.

La lógica es sencilla:

Prevenir y detectar temprano puede costar dinero, pero corregir un problema después de la entrega puede costar mucho más.

Por eso, la inversión en calidad debe verse como una inversión estratégica, buscando un equilibrio entre el costo de prevenir y el costo de corregir.

3. Reservas del proyecto

Las reservas funcionan como un colchón financiero para enfrentar situaciones que pueden afectar el presupuesto.

Reserva de contingencia

Se utiliza para riesgos conocidos, es decir, riesgos que fueron identificados durante la planificación y cuyo posible impacto podemos estimar.

Por ejemplo:

Sabemos que existe el riesgo de que una actividad se retrase y genere un costo adicional, por lo que reservamos dinero para responder a ese riesgo.

Reserva de gestión

Está relacionada con situaciones o riesgos no previstos o desconocidos que pueden aparecer durante el proyecto.

A diferencia de la contingencia, esta reserva no forma parte de la línea base de costos.

El uso de las reservas debe seguir las reglas de gobierno y aprobación establecidas para el proyecto.

4. Construcción del presupuesto

El presupuesto se construye paso a paso:

Estimar → Agrupar → Reservar → Aprobar → Controlar

Estimar

Se calcula el costo de cada actividad individual.

Agrupar

Se suman los costos de las actividades para obtener el presupuesto de cada fase y posteriormente el costo total del proyecto.

Reservar

Se incorpora la reserva de contingencia necesaria para responder a los riesgos identificados.

Establecer la línea base

Finalmente, se obtiene un presupuesto aprobado que servirá como referencia para controlar el desempeño financiero del proyecto.

5. Línea base presupuestal

La línea base de costos es la versión aprobada del presupuesto que se utiliza como referencia para medir y controlar los costos del proyecto.

Puede distribuirse por fases y periodos de tiempo.

La línea base incluye:

Estimaciones de costos aprobadas.

Presupuesto distribuido por fases.

Reserva de contingencia.

La línea base excluye la reserva de gestión.

Una vez aprobada, no debería modificarse simplemente porque los gastos estén aumentando. Los cambios deben pasar por un proceso formal de control de cambios.

Por eso, la línea base permite comparar:

Lo que habíamos planeado gastar vs. lo que realmente estamos gastando.

6. Curva S

La curva S es una gráfica que muestra cómo se va acumulando el presupuesto durante un periodo determinado.

Permite visualizar cómo debería distribuirse el gasto a lo largo del tiempo y detectar desviaciones.

Por ejemplo:

Mes 1: el gasto planificado está por debajo del límite aprobado.

Mes 2: el gasto aumenta y supera lo que estaba previsto para ese periodo.

Esto no significa automáticamente que el proyecto esté mal. Lo importante es analizar por qué ocurrió la desviación y qué impacto tendrá sobre el presupuesto y el flujo de caja.

7. Reprogramación y equilibrio de gastos

Si determinadas actividades están generando un gasto mayor al esperado, se puede analizar la posibilidad de reprogramar tareas y distribuir mejor los gastos en el tiempo, siempre respetando las reglas de control del proyecto.

El objetivo es mantener un equilibrio entre:

Actividades.

Gastos.

Presupuesto aprobado.

Flujo de caja.

Liquidez del proyecto.

No se trata simplemente de gastar menos, sino de administrar los recursos de manera que el proyecto pueda continuar funcionando sin comprometer su capacidad financiera.

8. Presupuesto detallado

El presupuesto detallado convierte la estimación general en un modelo financiero y técnico que permite tomar decisiones.

Permite conocer:

Qué actividades generan los costos.

Qué recursos se necesitan.

Cuánto cuesta cada fase.

Cuándo se producirán los gastos.

Qué riesgos pueden generar costos adicionales.

Cuánto dinero está reservado para contingencias.

Cuál es el presupuesto total aprobado.

También proporciona mayor claridad a directivos, equipos de proyecto y clientes, porque permite explicar de dónde salen los números y cómo se controlará el dinero.

9. El blindaje financiero del proyecto

Cuando tenemos:

Estimaciones + presupuesto detallado + reservas + línea base + control de cambios + seguimiento mediante la curva S

el proyecto tiene herramientas para controlar financieramente su ejecución.

Esto nos lleva a la pregunta central:

¿El proyecto vuela a ciegas o tiene blindaje de línea base?

Un proyecto que no conoce sus costos, no identifica sus riesgos y no tiene un presupuesto de referencia tiene dificultades para saber si está cumpliendo financieramente.

En cambio, una línea base permite tener un punto de referencia para comparar, detectar desviaciones y tomar decisiones con información.

Por eso, la línea base no es simplemente un documento financiero: es una herramienta de control para proteger el valor y la sostenibilidad del proyecto.
